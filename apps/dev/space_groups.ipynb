# Space groups: what they do and which reflections they remove (local test)

Two views of one space group, chosen for diffraction users rather than crystallographers:

1. **Reflections, allowed and why not.** A layer of the reciprocal lattice in its true geometry, with every
   systematically absent reflection marked by the symmetry that removes it (lattice centring, screw axis,
   glide plane), and a phase diagram showing *why*: the symmetry copies of each atom scatter into that
   reflection with phases that add to zero.
2. **Real vs reciprocal axes.** a, b, c and a\*, b\*, c\* together in 3D, and how far the reflection
   (h k l) is from the real direction [h k l] with the same digits.

For the symmetry itself in real space (symmetry elements, Wyckoff positions, an atom's copies) use the [Jmol Space Group Symmetry Visualizer](https://spacegroups.symotter.org/); it and other references are on the last tab.
