# THC accessible diffraction angles

For a given reflection's required detector angles ``(gamma, delta)`` (see
the [diffraction reflections](diffraction_reflections.ipynb) app), this
checks whether Bernina's optional THC sample-chamber hardware can actually
reach them. Each of the three detector-arm options physically limits the
accessible ``(gamma, delta)`` range:

- **In-vacuum detector** — a rectangular in-vacuum detector on a manually
  rotated flange.
- **Be window** — scattering out through a circular beryllium window on the
  rotatable top flange (dashed grey = the CF250 flange opening itself, the
  hard outer limit).
- **Otti** — the motorized in-vacuum arm. Its envelope comes from CAD/
  raytracing in the original tool rather than a closed-form aperture
  projection, so it's shown precomputed (`data/otti.npy`) rather than
  recomputed live.

Move a rotation slider to see the corresponding aperture position (top row)
and how the accessible angle envelope (bottom row, solid curve) sits within
the widest envelope reachable over all rotations (dashed curve).

Ported from the original
[`thc_angles`](https://github.com/swissfel/thc_angles) Binder app
(`thc_angles_2.ipynb`). The annotated hardware photos from that version are
not reproduced here — they hot-link an external, personally hosted image
store, which doesn't fit a fast, offline-friendly Binder build — but the
geometry math and interactive sliders are otherwise unchanged.
