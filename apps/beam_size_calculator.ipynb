# Beam size / divergence calculator

Enter a measured FWHM spot size at a known distance from focus for the
horizontal (X) and vertical (Y) axes; the tool infers the focus waist and
Rayleigh range (assuming diffraction-limited Gaussian propagation) and plots
the beam envelope alongside named beamline components. Component distances
default to `data/beamline.csv`; add/remove rows to try other layouts.

```python
import sys
from pathlib import Path

try:
    import bernina_prep  # noqa: F401
except ImportError:
    sys.path.insert(0, str(Path.cwd().parent))

%matplotlib widget
from IPython.display import display

from bernina_prep.beamsize import GeometricBeamApp

repo_root = Path.cwd().parent
beamline_csv = repo_root / "data" / "beamline.csv"

app = GeometricBeamApp(beamline_csv=beamline_csv)
display(app.ui)
```
