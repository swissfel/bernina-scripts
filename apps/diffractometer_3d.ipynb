# Mock diffractometer (3D, You geometry)

Interactive three.js model of Bernina's six-circle diffractometer (H. You,
J. Appl. Cryst. 1999, 32, 614-623): mu, eta, chi, phi on the sample, gamma and
delta on the detector, the geometry
[`diffraction_reflections.ipynb`](diffraction_reflections.ipynb) solves.
Drag to orbit, scroll to zoom, move the circles with the sliders. Axes are
SwissFEL/Bernina's: z along the beam, y up, x to the left looking along the
beam. The page is `bernina_prep/diffractometer_3d.html` (self-contained,
works offline); the reflections app embeds the same page.
